```python
"""
Student Marks Range Counter
---------------------------

Objective:
    Apply loops to realistic data‑processing scenarios.

Task:
    Given a list of student marks, count how many students scored:
        - 90+
        - 75–89
        - 50–74
        - below 50

Approach:
    - Initialize counters for each category.
    - Loop through each mark.
    - Use conditional checks to increment the correct counter.
"""


def count_mark_ranges(marks: list[int]) -> dict[str, int]:
    """
    Count students in different mark ranges.

    Parameters:
        marks (list[int]): List of student marks.

    Returns:
        dict[str, int]: A dictionary containing counts for each category.

    Explanation:
        - We manually check each mark using if/elif conditions.
        - Increment the appropriate counter based on the range.
    """

    # Initialize counters
    count_90_plus = 0
    count_75_89 = 0
    count_50_74 = 0
    count_below_50 = 0

    # Loop through each mark and categorize
    for score in marks:
        if score >= 90:
            count_90_plus += 1
        elif 75 <= score <= 89:
            count_75_89 += 1
        elif 50 <= score <= 74:
            count_50_74 += 1
        else:
            count_below_50 += 1

    # Return results in a clean dictionary
    return {
        "90_plus": count_90_plus,
        "75_to_89": count_75_89,
        "50_to_74": count_50_74,
        "below_50": count_below_50
    }


# ---------------------------------------------------------
# Demonstration
# ---------------------------------------------------------

marks = [78, 92, 45, 67, 88, 53, 99]

result = count_mark_ranges(marks)

print("Students scoring 90+     :", result["90_plus"])
print("Students scoring 75–89   :", result["75_to_89"])
print("Students scoring 50–74   :", result["50_to_74"])
print("Students below 50        :", result["below_50"])
```
