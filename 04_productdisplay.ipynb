```python
text = input("Enter a string: ")

frequency = {}   # dictionary to store character counts

for char in text:          # look at each character one by one
    if char in frequency:  # if we have seen this character before
        frequency[char] += 1   # increase its count
    else:
        frequency[char] = 1    # first time seeing this character

# print each character and its count
for char, count in frequency.items():
    print(char, "->", count)
```

```python
#Find the second-largest number in a list without using sort().
numbers = [12, 45, 7, 89, 23, 67]

largest = float('-inf')        # start with very small number
second_largest = float('-inf') # same for second largest

for num in numbers:
    # If current number is bigger than largest, update both
    if num > largest:
        second_largest = largest
        largest = num

    # If current number is between largest and second largest
    elif num > second_largest and num != largest:
        second_largest = num

print("Second largest number is:", second_largest)
```

```python
text = input("Enter a string: ")

is_palindrome = True   # assume it IS a palindrome first

# Loop from start to middle of the string
for i in range(len(text) // 2):
    # Compare character from front and corresponding character from back
    if text[i] != text[-(i + 1)]:
        is_palindrome = False   # found a mismatch
        break                   # no need to check further

if is_palindrome:
    print("Palindrome")
else:
    print("Not a palindrome")
```
