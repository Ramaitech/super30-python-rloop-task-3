```python
"""
Transaction Total Calculator
----------------------------

Objective:
    Apply loops to realistic data‑processing scenarios.

Task:
    Given a list of transaction amounts, calculate the total value
    WITHOUT using Python's built‑in sum() function.

Approach:
    - Initialize a running total at 0.
    - Loop through each transaction amount.
    - Add each amount to the running total.
    - Return the final accumulated total.
"""


def calculate_transaction_total(transactions: list[int]) -> int:
    """
    Manually compute the total value of all transactions.

    Parameters:
        transactions (list[int]): A list of integer transaction amounts.

    Returns:
        int: The total of all transaction values.

    Explanation:
        We avoid using sum() to demonstrate how accumulation works internally.
        A loop processes each element and updates the running total.
    """

    total = 0  # Start with zero; this variable will accumulate the sum

    # Loop through each transaction amount
    for amount in transactions:
        total += amount  # Add the current amount to the running total

    return total  # Final total after processing all transactions


# ---------------------------------------------------------
# Demonstration
# ---------------------------------------------------------

transactions = [1200, 450, 800, 1500, 2300, 700, 100]

total_value = calculate_transaction_total(transactions)

print("Total transaction value:", total_value)
```

```python
"""
Highest & Lowest Transaction Finder
-----------------------------------

Objective:
    Apply loops to realistic data‑processing scenarios.

Task:
    From the given list of transactions, find:
        - Highest transaction
        - Lowest transaction
    WITHOUT using Python's built‑in max() or min().

Approach:
    - Initialize highest and lowest with the first element.
    - Loop through each transaction.
    - Update highest if a larger value is found.
    - Update lowest if a smaller value is found.
"""


def find_high_low(transactions: list[int]) -> tuple[int, int]:
    """
    Determine the highest and lowest transaction values manually.

    Parameters:
        transactions (list[int]): List of transaction amounts.

    Returns:
        tuple[int, int]: (highest_value, lowest_value)

    Explanation:
        - Start by assuming the first element is both highest and lowest.
        - Compare each element with current highest and lowest.
        - Update accordingly.
    """

    # Initialize both highest and lowest with the first transaction
    highest = transactions[0]
    lowest = transactions[0]

    # Loop through each transaction
    for amount in transactions:
        # Update highest if current amount is greater
        if amount > highest:
            highest = amount

        # Update lowest if current amount is smaller
        if amount < lowest:
            lowest = amount

    return highest, lowest


# ---------------------------------------------------------
# Demonstration
# ---------------------------------------------------------

transactions = [1200, 450, 800, 1500, 2300, 700, 100]

highest, lowest = find_high_low(transactions)

print("Highest transaction:", highest)
print("Lowest transaction:", lowest)
```
