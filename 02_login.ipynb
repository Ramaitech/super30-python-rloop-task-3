```python
correct_password = "super30"   # the actual password

attempts = 0                   # how many times user has tried

while attempts < 3:            # allow only 3 attempts
    entered = input("Enter password: ")

    if entered == correct_password:
        print("Login successful.")
        break                  # stop the loop after success
    else:
        attempts += 1          # increase attempt count
        print("Incorrect password.")

        # show remaining attempts
        print("Attempts left:", 3 - attempts)

# if attempts reached 3 and user never logged in
if attempts == 3:
    print("Maximum attempts reached. System locked.")
```
